```python
import kagglehub, shutil, hashlib
from pathlib import Path
import step0_unsw as s0
src = Path(kagglehub.dataset_download("mrwellsdavid/unsw-nb15"))
dst = Path("/content/unsw_std"); dst.mkdir(exist_ok=True)
# Bản Kaggle đặt ngược tên: file "testing-set" mới là tập train chuẩn (175.341 dòng)
shutil.copy(next(src.rglob("UNSW_NB15_testing-set.csv")), dst / "UNSW_NB15_training-set.csv")
shutil.copy(next(src.rglob("UNSW_NB15_training-set.csv")), dst / "UNSW_NB15_testing-set.csv")
tr = s0.read_split(dst / "UNSW_NB15_training-set.csv")
te = s0.read_split(dst / "UNSW_NB15_testing-set.csv")
cols = [c for v in s0.VIEWS.values() for c in v] + ["attack_cat"]
fp = hashlib.sha1(s0.row_hash(tr, cols).tobytes() + s0.row_hash(te, cols).tobytes()).hexdigest()[:16]
print(len(tr), len(te), fp, "-> KHỚP step 0" if fp == "aa6393174bdb3971" else "-> KHÁC step 0, dừng lại")
```
